### Explanations 

As the LLMs aren't able to understand raw texts, we need to convert them into numbers. To do so, the first step is to split the text into a smaller version so the model will be able to convert it afterward.

The tokenization will decompose onto tokens the sentence that we have in entry. A tokens is a part of the text, that is understandable for the LLM.

Regarding the implementation, a tokenizer go through regular steps : 

1. Normalization : Convert text into standard, machine-friendly form before being broken into tokens 
2. Pre-tokenization : input is broken into simple chunks before implementing any complex tokenizations algorithms
3. Subword segmentation : words are broken into meaningful arguments
4. Mapping to IDs : mapped the tokens to unique, numerical IDs
5. Detokenization and potential pitfalls : converting tokens to their original form, the human text

Tokenization algorithms : 
- **BPE** : most frequent pairs of characters or bytes
- **WordPiece** : merge individual characters into tokens based on probability 
- **Unigram** : Probabilistic model where every token is assigned a possibility score
- **SentencePiece** : versatile toolkit that support both BPE and unigram ; tokenizes raw text as a seamless stream

*How to correctly chose the tokenizer algorithm ?*
- if you use a LLM model, chose the tokenizer that they are using
- Compare the families : today, the mostly used algorithms are SentencePiece and BPE
- Things to take in considerations when you chose : 
    1. length of the token
    2. the language
    3. are you using a programation oriented LLM ?
    4. length of the vocabulary
    5. metric - compression rate : nb_token/nb_caracters or nb_token/nb_word
